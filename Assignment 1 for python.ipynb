## Q1

- Take the customer's age as input.
- Use if, elif and else to check the age group.
- Print the correct ticket price for that age.
- This program uses conditional statements to select one ticket price based on the customer's age.


```python
age = int(input("Enter your age: "))
if age < 5:
    print("Ticket is Free")
elif age <= 17:
    print("Your ticket price is Rs. 100")
elif age <= 59:
    print("Your ticket price is Rs. 200")
else:
    print("Your ticket price is Rs. 120")
```

## Q2

- Take the number of units consumed as input.
- Charge the first 100 units at Rs. 3, the next 100 at Rs. 5, and units above 200 at Rs. 8.
- Print the total electricity bill.
- This program uses conditions to calculate the electricity bill according to the usage slabs given in the assignment.


```python
units = int(input("Enter units consumed: "))
if units <= 100:
    bill = units * 3
elif units <= 200:
    bill = (100 * 3) + ((units - 100) * 5)
else:
    bill = (100 * 3) + (100 * 5) + ((units - 200) * 8)
print("Total bill: Rs.", bill)
```

## Q4

- Start total savings and number of days at 0.
- Use a while loop to keep asking for daily savings.
- Stop when the total reaches or crosses Rs. 1000.
- Print how many days it took and the final total.
- The while loop continues until Riya reaches the savings goal of Rs. 1000.


```python
total = 0
days = 0
while total < 1000:
    savings = int(input("Enter today's savings: "))
    total = total + savings
    days = days + 1
print("Goal reached in", days, "days! Total saved: Rs.", total)
```

## Q5

- Store the correct PIN as 4321.
- Give the user only three attempts.
- Print Access granted for the correct PIN and Wrong PIN for an incorrect PIN.
- After three wrong attempts, print Card blocked.
- This program uses a while loop and conditions to limit the user to three PIN attempts.


```python
correct_pin = 4321
attempts = 0
while attempts < 3:
    pin = int(input("Enter PIN: "))
    if pin == correct_pin:
        print("Access granted")
        break
    else:
        print("Wrong PIN")
        attempts = attempts + 1
if attempts == 3:
    print("Card blocked")
```

## Q6

- Take N as input.
- Use a for loop to check every number from 1 to N.
- Use % 2 to determine whether each number is even or odd.
- Count and print the number of even and odd values.
- A number with remainder 0 when divided by 2 is even; otherwise it is odd.


```python
n = int(input("Enter N: "))
even_count = 0
odd_count = 0
for i in range(1, n + 1):
    if i % 2 == 0:
        even_count = even_count + 1
    else:
        odd_count = odd_count + 1
print("Even numbers:", even_count)
print("Odd numbers:", odd_count)
```

## Q7

- Store 27 as the secret number.
- Use a while loop to keep asking for a guess.
- Print Too high or Too low after each wrong guess.
- When the answer is correct, print the number of guesses.
- The loop continues until the user guesses 27, while a counter records the number of attempts.


```python
secret = 27
guess = 0
count = 0
while guess != secret:
    guess = int(input("Guess the number: "))
    count = count + 1
    if guess > secret:
        print("Too high")
    elif guess < secret:
        print("Too low")
    else:
        print("Correct! You took", count, "guesses.")
```

## Q8

- Take marks out of 100 as input.
- First check whether the marks are invalid.
- Use if and elif to assign A, B, C, D or Fail.
- Print Invalid marks for values below 0 or above 100.
- The conditions are checked from the invalid range and highest grade down to the failing range.


```python
marks = int(input("Enter marks: "))
if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
elif marks >= 35:
    print("Grade: D")
else:
    print("Fail")
```

## Q9

- Take the bill number as an integer.
- Use % 10 to get the last digit and // 10 to remove it.
- Use a while loop to add all the digits.
- Give a free gift if the sum is more than 20.
- The while loop processes the bill number one digit at a time and calculates the total of its digits.


```python
bill_number = int(input("Enter bill number: "))
number = bill_number
digit_sum = 0
while number > 0:
    digit = number % 10
    digit_sum = digit_sum + digit
    number = number // 10
print("Sum of digits:", digit_sum)
if digit_sum > 20:
    print("Congratulations! You win a free gift.")
else:
    print("No free gift.")
```

## Q10

- Take the number of steps as input.
- Use an outer for loop to create each row.
- Use an inner for loop to print the stars.
- Increase the number of stars on every new row.
- This program uses nested for loops: the outer loop controls the rows and the inner loop prints the stars.


```python
steps = int(input("Enter number of steps: "))
for i in range(1, steps + 1):
    for j in range(i):
        print("*", end=" ")
    print()
```
