## Q1 - Phone Contact Book

- Create a dictionary for contacts.
- Use a while-loop menu: Add, Search, Show All, Exit.
- Check duplicate and missing names.
- Uses a dictionary, while loop and if-else.


```python
contacts = {}
choice = ""
while choice != "4":
    print("1. Add contact\n2. Search contact\n3. Show all contacts\n4. Exit")
    choice = input("Enter choice: ")
    if choice == "1":
        name = input("Enter name: ")
        if name in contacts:
            print("Contact already exists")
        else:
            contacts[name] = input("Enter phone number: ")
    elif choice == "2":
        name = input("Enter name to search: ")
        if name in contacts:
            print("Phone number:", contacts[name])
        else:
            print("Contact not found")
    elif choice == "3":
        for name, phone in contacts.items():
            print(name, ":", phone)
```

## Q3 - Student Marks Dictionary

- Use the given marks dictionary.
- Print Pass/Fail using 40 as pass mark.
- Find the topper and create a passed-students dictionary.
- Uses dictionary items and a loop.


```python
marks = {"Aman": 82, "Priya": 91, "Rahul": 45, "Sneha": 67, "Kiran": 30}
passed_students = {}
topper = ""
highest_marks = 0
for name, mark in marks.items():
    if mark >= 40:
        print(name, "- Pass")
        passed_students[name] = mark
    else:
        print(name, "- Fail")
    if mark > highest_marks:
        highest_marks = mark
        topper = name
print("Topper:", topper, highest_marks)
print("Passed students:", passed_students)
```

## Q4 - Common Friends

- Store both friend lists as sets.
- Use intersection for mutual friends.
- Use difference for Riya-only and suggestions.
- Use union for all friends.


```python
riya_friends = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}
print("Mutual friends:", riya_friends.intersection(arjun_friends))
print("Friends only Riya has:", riya_friends - arjun_friends)
print("All friends:", riya_friends.union(arjun_friends))
print("Friend suggestions for Riya:", arjun_friends - riya_friends)
```

## Q5 - Character Counter

- Take text as input.
- Ignore spaces and count characters in a dictionary.
- Find the most frequent character.
- The dictionary stores each character and its count.


```python
text = input("Enter text: ")
counts = {}
for char in text:
    if char != " ":
        if char in counts:
            counts[char] = counts[char] + 1
        else:
            counts[char] = 1
most_char = ""
most_count = 0
for char, count in counts.items():
    if count > most_count:
        most_char = char
        most_count = count
print(counts)
print("Most frequent character:", most_char, "("+str(most_count)+" times)")
```

## Q6 - Restaurant Menu Billing

- Use the given menu dictionary.
- Keep taking orders until bill is typed.
- Add valid item prices; report unavailable items.
- Apply 10% discount if total is above Rs. 200.


```python
menu = {"dosa": 60, "idli": 40, "vada": 30, "coffee": 25, "tea": 15}
total = 0
item = input("Enter item: ").lower()
while item != "bill":
    if item in menu:
        total = total + menu[item]
    else:
        print("Item not available")
    item = input("Enter item: ").lower()
if total > 200:
    total = total - (total * 0.10)
    print("10% discount applied")
print("Total bill: Rs.", total)
```

## Q7 - City Coordinates (Tuples)

- Use the given city tuples.
- Unpack city, latitude and longitude in a loop.
- Find the highest latitude.
- Tuples are immutable, so their values cannot be changed.


```python
cities = [("Hyderabad", 17.38, 78.48), ("Mumbai", 19.07, 72.87),
          ("Chennai", 13.08, 80.27), ("Delhi", 28.61, 77.20)]
north_city = ""
highest_latitude = 0
for city, latitude, longitude in cities:
    print(city, latitude, longitude)
    if latitude > highest_latitude:
        highest_latitude = latitude
        north_city = city
print("Farthest north city:", north_city)
# cities[1][1] = 20.00
# TypeError: tuples are immutable and cannot be changed.
```

## Q8 - Library Book Tracker

- Use the given books dictionary.
- Ask for a book name.
- Check missing, unavailable, or available.
- Reduce available copies by 1 when issued.


```python
books = {"Python Basics": 3, "Data Science 101": 0, "Machine Learning": 2, "SQL Guide": 1}
book_name = input("Enter book name to borrow: ")
if book_name not in books:
    print("Book not found")
elif books[book_name] == 0:
    print("Currently unavailable")
else:
    books[book_name] = books[book_name] - 1
    print("Book issued successfully")
print("Updated books:", books)
```

## Q9 - Group Employees by Department

- Use the given employee tuples.
- Create a dictionary with departments as keys.
- Store employee names in department lists.
- Print employee count for each department.


```python
employees = [("Ravi", "IT"), ("Anjali", "HR"), ("Suresh", "IT"),
             ("Meena", "Finance"), ("Arun", "HR"), ("Divya", "IT")]
departments = {}
for name, department in employees:
    if department in departments:
        departments[department].append(name)
    else:
        departments[department] = [name]
print(departments)
for department, names in departments.items():
    print(department, ":", len(names), "employees")
```

## Q10 - Quiz Answer Checker

- Use the answer-key tuple.
- Ask for five answers one at a time.
- Store wrong question numbers in a set.
- Print score and the required remark.


```python
answer_key = ("B", "C", "A", "D", "B")
wrong_questions = set()
score = 0
for i in range(5):
    answer = input("Enter answer for question " + str(i + 1) + ": ").upper()
    if answer == answer_key[i]:
        print("Question", i + 1, "is correct")
        score = score + 1
    else:
        print("Question", i + 1, "is wrong")
        wrong_questions.add(i + 1)
print("Wrong question numbers:", wrong_questions)
print("Final score:", score, "out of 5")
if score == 5:
    print("Excellent")
elif score >= 3:
    print("Good")
else:
    print("Needs improvement")
```
