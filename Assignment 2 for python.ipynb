## Q1 - Class Marks Report

- Use the given list of marks for 8 students.
- Find the highest marks, lowest marks, and class average.
- Use a loop to count how many students scored 35 or more.
- This program uses a list and basic calculations to create a simple class marks report.


```python
marks = [78, 45, 92, 33, 67, 88, 51, 29]
highest = max(marks)
lowest = min(marks)
average = sum(marks) / len(marks)
passed = 0
for mark in marks:
    if mark >= 35:
        passed = passed + 1
print("Highest marks:", highest)
print("Lowest marks:", lowest)
print("Class average:", average)
print("Students passed:", passed)
```

## Q2 - Grocery Shopping List

- Start with an empty shopping list.
- Keep asking for items until the user types done.
- Do not add an item if it is already in the list.
- Print each item with a number and the total number of items.
- This program uses a while loop and a list to collect unique shopping items.


```python
shopping_list = []
item = input("Enter item: ")
while item.lower() != "done":
    if item in shopping_list:
        print("Already added")
    else:
        shopping_list.append(item)
    item = input("Enter item: ")
for i in range(len(shopping_list)):
    print(i + 1, ".", shopping_list[i])
print("Total items:", len(shopping_list))
```

## Q3 - Username Validator

- Take a username from the user.
- Check that its length is between 5 and 12 characters.
- Check that it has no spaces and does not start with a digit.
- This program uses string checks and if-elif conditions to validate the username.


```python
username = input("Enter username: ")
if len(username) < 5 or len(username) > 12:
    print("Invalid: Username must be between 5 and 12 characters")
elif " " in username:
    print("Invalid: Username cannot contain spaces")
elif username[0].isdigit():
    print("Invalid: Username cannot start with a digit")
else:
    print("Valid username")
```

## Q4 - Vowel Counter for a Tweet

- Take a sentence from the user.
- Loop through every character.
- Count vowels, consonants, and spaces.
- Ignore digits and special characters for vowel/consonant count.
- This program checks each character and keeps separate counts.


```python
sentence = input("Enter a sentence: ")
vowels = 0
consonants = 0
spaces = 0
for char in sentence:
    if char.lower() in "aeiou":
        vowels = vowels + 1
    elif char.isalpha():
        consonants = consonants + 1
    elif char == " ":
        spaces = spaces + 1
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Spaces:", spaces)
```

## Q5 - Palindrome Name Checker

- Take the player's name as input.
- Convert it to lowercase so letter case does not matter.
- Compare the name with its reversed form.
- This program checks whether the name reads the same forwards and backwards.


```python
name = input("Enter your name: ")
name_lower = name.lower()
if name_lower == name_lower[::-1]:
    print("Bonus! Your name is a palindrome.")
else:
    print("Your name is not a palindrome.")
```

## Q6 - Temperature Tracker

- Use the given temperature and day lists.
- Print days with temperature above 37.
- Find the hottest day and temperature.
- Create a new list of Fahrenheit temperatures.
- This program uses lists and loops to analyse the weekly temperatures.


```python
temps = [32, 35, 38, 41, 36, 29, 40]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
print("Heat alert days:")
for i in range(len(temps)):
    if temps[i] > 37:
        print(days[i])
hottest_temp = max(temps)
hottest_index = temps.index(hottest_temp)
print("Hottest day:", days[hottest_index])
print("Hottest temperature:", hottest_temp)
fahrenheit = []
for temp in temps:
    f = temp * 9 / 5 + 32
    fahrenheit.append(f)
print("Temperatures in Fahrenheit:", fahrenheit)
```

## Q7 - Word Frequency in a Sentence

- Take a sentence and split it into words.
- Find the longest and shortest words.
- Count words that start with a capital letter.
- This program loops through the words to compare their lengths and first letters.


```python
sentence = input("Enter a sentence: ")
words = sentence.split()
longest = words[0]
shortest = words[0]
capital_count = 0
for word in words:
    if len(word) > len(longest):
        longest = word
    if len(word) < len(shortest):
        shortest = word
    if word[0].isupper():
        capital_count = capital_count + 1
print("Longest word:", longest)
print("Shortest word:", shortest)
print("Words starting with capital letter:", capital_count)
```

## Q8 - Remove Negative Balances

- Use the given list of balances.
- Create a positive_accounts list for balances above 0.
- Create a negative_accounts list for balances below 0.
- Print both lists and the total amount owed.
- This program separates balances into two new lists using a loop.


```python
balances = [1500, -200, 3000, -50, 0, 750, -1000]
positive_accounts = []
negative_accounts = []
for balance in balances:
    if balance > 0:
        positive_accounts.append(balance)
    elif balance < 0:
        negative_accounts.append(balance)
total_owed = 0
for balance in negative_accounts:
    total_owed = total_owed + balance
print("Positive accounts:", positive_accounts)
print("Negative accounts:", negative_accounts)
print("Total amount owed:", abs(total_owed))
```

## Q9 - Password Strength Checker

- Take a password as input.
- Check each character for uppercase, lowercase, and digit.
- Check whether it has at least 8 characters.
- Print Strong, Medium, or Weak based on conditions met.
- This program counts how many of the four password rules are satisfied.


```python
password = input("Enter password: ")
has_upper = False
has_lower = False
has_digit = False
for char in password:
    if char.isupper():
        has_upper = True
    elif char.islower():
        has_lower = True
    elif char.isdigit():
        has_digit = True
conditions = 0
if has_upper:
    conditions = conditions + 1
if has_lower:
    conditions = conditions + 1
if has_digit:
    conditions = conditions + 1
if len(password) >= 8:
    conditions = conditions + 1
if conditions == 4:
    print("Strong")
elif conditions == 2 or conditions == 3:
    print("Medium")
else:
    print("Weak")
```

## Q10 - Simple Name Initials Generator

- Take the full name and split it into separate names.
- Create uppercase initials with dots.
- Print the full name in title case.
- Print initials for earlier names and the surname in full.
- This program uses string methods and loops to format the name in three ways.


```python
full_name = input("Enter full name: ")
names = full_name.split()
initials = ""
for name in names:
    initials = initials + name[0].upper() + "."
print("Initials:", initials)
print("Title case:", full_name.title())
surname_format = ""
for i in range(len(names) - 1):
    surname_format = surname_format + names[i][0].upper() + ". "
surname_format = surname_format + names[-1].title()
print("Surname format:", surname_format)
```
